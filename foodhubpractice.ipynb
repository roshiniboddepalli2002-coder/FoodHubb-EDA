```python
def calc_revenue(row):
    cost = row['cost_of_the_order']

    if cost > 20:
        return cost * 0.25

    elif cost > 5:
        return cost * 0.15

    else:
        return 0
```

```python
df['commission'] = df.apply(calc_revenue, axis=1)
```

```python
total_revenue = df['commission'].sum()
print(total_revenue)
```

```python
df['total_time'] = df['food_preparation_time'] + df['delivery_time']
```
